# STEP 8 — CREATE THE FOOD DIVERSITY FUNCTION

```python
def calculate_diversity_score(person):
    score = 0

    if person["fruit_servings"] > 0:
        score += 1

    if person["vegetable_servings"] > 0:
        score += 1

    if person["whole_grain_servings"] > 0:
        score += 1

    if person["protein_servings"] > 0:
        score += 1

    return score
```

# STEP 13 — FRUIT/VEGETABLE SIGNAL

```python
def get_fruit_vegetable_signal(person):
    total = person["fruit_servings"] + person["vegetable_servings"]

    if total <= 1:
        return "LOW CONSUMPTION SIGNAL"
    else:
        return "BETTER CONSUMPTION SIGNAL"
```

# STEP 14 — FRIED-FOOD SIGNAL

```python
def get_fried_food_signal(person):
    if person["fried_food_servings"] >= 3:
        return "HIGH"
    else:
        return "LOWER"
```

# STEP 15 — CREATE THE RISK SCORE

```python
def calculate_risk_score(person):
    score = 0

    diversity = calculate_diversity_score(person)

    if diversity <= 2:
        score += 2

    if person["processed_food_servings"] >= 3:
        score += 2

    if person["sugary_drink_servings"] >= 2:
        score += 2

    fruit_vegetable_total = (person["fruit_servings"] + person["vegetable_servings"])

    if fruit_vegetable_total <= 1:
        score += 2

    if person["fried_food_servings"] >= 3:
        score += 1

    return score
```

# STEP 16 — CLASSIFY THE RISK SCORE

```python
def get_risk_level(score):
    if score <= 2:
        return "LOW RISK SIGNAL"

    elif score <= 5:
        return "MODERATE RISK SIGNAL"

    else:
        return "HIGHER RISK SIGNAL"
```

# STEP 23 — VALIDATE THE DATA

```python
valid_age_groups = [
    "18-29",
    "30-44",
    "45-59",
    "60+"
]

numeric_fields = [
    "fruit_servings",
    "vegetable_servings",
    "whole_grain_servings",
    "protein_servings",
    "processed_food_servings",
    "sugary_drink_servings",
    "fried_food_servings",
    "water_servings",
    "meal_frequency"
]

required_fields = ["person_id", "age_group", "location"] + numeric_fields


def validate_record(person, food_data=None):
    errors = []

    for field in required_fields:
        if field not in person:
            errors.append(f"Missing field: {field}")

    if errors:
        return errors

    if person["person_id"].strip() == "":
        errors.append("Participant ID cannot be blank.")

    if person["location"].strip() == "":
        errors.append("Location cannot be blank.")

    if person["age_group"] not in valid_age_groups:
        errors.append("Invalid age group.")

    for field in numeric_fields:
        value = person[field]

        if isinstance(value, bool) or not isinstance(value, (int, float)):
            errors.append(f"{field} must contain a number.")
        elif value < 0:
            errors.append(f"{field} cannot be negative.")

    if food_data is not None:
        existing_ids = [item["person_id"] for item in food_data]
        if person["person_id"] in existing_ids:
            errors.append("Participant ID already exists.")

    return errors


# isinstance() checks whether a value belongs to a particular data type.
# For example, it can check if a value is an integer, float, or string.
# In this project, it can be used to make sure that values entered by the user are the correct type before we perform calculations.

# append() adds a new item to the end of a list.
# In this project, append() is used to add error messages to the errors list when a record fails validation.
```

# STEP 24 — VALIDATE THE AGE GROUP

# STEP 25 — VALIDATE THE ENTIRE DATASET

```python
def validate_dataset(food_data):
    invalid_count = 0

    for person in food_data:
        errors = validate_record(person)

        if errors:
            invalid_count += 1

            print("\nInvalid record:", person["person_id"])

            for error in errors:
                print("-", error)

    print("\nInvalid records:", invalid_count)

validate_dataset(food_data)
```

# STEP 26 — CHECK DUPLICATE IDs

```python
def check_duplicate_ids(food_data):
    ids = []
    duplicates = []

    for person in food_data:
        person_id = person["person_id"]

        if person_id in ids:
            duplicates.append(person_id)
        else:
            ids.append(person_id)

    return duplicates

duplicates = check_duplicate_ids(food_data)

if duplicates:
    print("Duplicate IDs:", duplicates)
else:
    print("No duplicate participant IDs found.")
```

# STEP 27 — LOCATION ANALYSIS

```python
def location_analysis(food_data):
    location_scores = {}
    location_counts = {}

    for person in food_data:
        location = person["location"]
        score = calculate_risk_score(person)

        if location not in location_scores:
            location_scores[location] = score
            location_counts[location] = 1

        else:
            location_scores[location] += score
            location_counts[location] += 1

    print("\n========== LOCATION SUMMARY ==========")

    for location in location_scores:
        average = (
            location_scores[location]
            / location_counts[location]
        )

        print(
            location,
            "Records:", location_counts[location],
            "Average Risk Score:", round(average, 2)
        )
```

# STEP 28 — RISK FACTOR FREQUENCY

```python
# Which dietary risk signal appears most frequently?

def risk_factor_frequency(food_data):

    diversity_risk = 0
    processed_risk = 0
    sugary_risk = 0
    fruit_vegetable_risk = 0
    fried_risk = 0

    for person in food_data:

        if calculate_diversity_score(person) <= 2:
            diversity_risk += 1

        if person["processed_food_servings"] >= 3:
            processed_risk += 1

        if person["sugary_drink_servings"] >= 2:
            sugary_risk += 1

        if (
            person["fruit_servings"]
            + person["vegetable_servings"]
        ) <= 1:
            fruit_vegetable_risk += 1

        if person["fried_food_servings"] >= 3:
            fried_risk += 1

    print("\n========== RISK FACTOR FREQUENCY ==========")

    print("Low Diversity:", diversity_risk)

    print("High Processed Food:", processed_risk)

    print("High Sugary Drinks:", sugary_risk)

    print("Low Fruit/Vegetable:", fruit_vegetable_risk)

    print("High Fried Food:", fried_risk)
```

# STEP 29 — RISK RANKING (FIND THE HIGHEST-RISK LOCATION)

```python
def risk_ranking(food_data):

    ranked_records = sorted(
        food_data,
        key=calculate_risk_score,
        reverse=True
    )

    print("\n========== RISK RANKING ==========")

    position = 1

    for person in ranked_records:

        print(
            position,
            person["person_id"],
            "-",
            person["location"],
            "- Risk Score:",
            calculate_risk_score(person)
        )

        position += 1
```

# STEP 30 — ADD A NEW PARTICIPANT 

```python
def get_non_negative_integer(prompt):

    while True:

        try:
            value = int(input(prompt))

            if value < 0:
                print("Value cannot be negative.")

            else:
                return value

        except ValueError:
            print("Please enter a valid number.")



def add_record(food_data):

    print("\n========== ADD PARTICIPANT ==========")

    person = {}

    person["person_id"] = input(
        "Enter Participant ID: "
    ).strip().upper()

    person["age_group"] = input(
        "Enter Age Group: "
    ).strip()

    person["location"] = input(
        "Enter Location: "
    ).strip()

    person["fruit_servings"] = get_non_negative_integer(
        "Fruit servings: "
    )

    person["vegetable_servings"] = get_non_negative_integer(
        "Vegetable servings: "
    )

    person["whole_grain_servings"] = get_non_negative_integer(
        "Whole grain servings: "
    )

    person["protein_servings"] = get_non_negative_integer(
        "Protein servings: "
    )

    person["processed_food_servings"] = get_non_negative_integer(
        "Processed food servings: "
    )

    person["sugary_drink_servings"] = get_non_negative_integer(
        "Sugary drink servings: "
    )

    person["fried_food_servings"] = get_non_negative_integer(
        "Fried food servings: "
    )

    person["water_servings"] = get_non_negative_integer(
        "Water servings: "
    )

    person["meal_frequency"] = get_non_negative_integer(
        "Meal frequency: "
    )

    errors = validate_record(person, food_data)

    if not errors:
        food_data.append(person)
        print("\nParticipant added successfully.")
    else:
        print("\nParticipant was not added.")
        for message in errors:
            print("-", message)
```

# STEP 31 — BUILD THE NESTED MENU

```python
def view_records(food_data):
    print("\n========== ALL PARTICIPANTS ==========")

    for person in food_data:
        print(
            person["person_id"],
            "|", person["age_group"],
            "|", person["location"],
            "| Fruit:", person["fruit_servings"],
            "| Vegetables:", person["vegetable_servings"],
            "| Processed:", person["processed_food_servings"]
        )


def show_risk_analysis(food_data):
    print("\n========== RISK ANALYSIS ==========")

    for person in food_data:
        score = calculate_risk_score(person)
        print(
            person["person_id"],
            "| Score:", score,
            "|", get_risk_level(score),
            "| Diversity:", calculate_diversity_score(person)
        )


def show_high_risk_records(food_data):
    print("\n========== HIGHER-RISK SIGNALS ==========")
    found = False

    for person in food_data:
        score = calculate_risk_score(person)

        if score >= 6:
            print(
                person["person_id"],
                "| Location:", person["location"],
                "| Score:", score,
                "|", get_risk_level(score)
            )
            found = True

    if not found:
        print("No higher-risk records found.")


def highest_risk_participant(food_data):
    highest_person = None
    highest_score = -1

    for person in food_data:
        score = calculate_risk_score(person)

        if score > highest_score:
            highest_score = score
            highest_person = person

    if highest_person is not None:
        print(
            "Highest Risk Participant:",
            highest_person["person_id"],
            "| Score:", highest_score,
            "| Location:", highest_person["location"]
        )

    return highest_person


def generate_summary(food_data):
    total = len(food_data)

    if total == 0:
        print("No records available.")
        return None

    scores = [calculate_risk_score(person) for person in food_data]
    diversity_scores = [calculate_diversity_score(person) for person in food_data]

    low_count = sum(1 for score in scores if score <= 2)
    moderate_count = sum(1 for score in scores if 3 <= score <= 5)
    high_count = sum(1 for score in scores if score >= 6)

    processed_count = sum(
        1 for person in food_data
        if person["processed_food_servings"] >= 3
    )

    sugary_count = sum(
        1 for person in food_data
        if person["sugary_drink_servings"] >= 2
    )

    print("\n========== OVERALL SUMMARY ==========")
    print("Total Participants:", total)
    print("Average Risk Score:", round(sum(scores) / total, 2))
    print("Highest Risk Score:", max(scores))
    print("Low Risk Records:", low_count, f"({low_count / total * 100:.1f}%)")
    print("Moderate Risk Records:", moderate_count, f"({moderate_count / total * 100:.1f}%)")
    print("Higher Risk Records:", high_count, f"({high_count / total * 100:.1f}%)")
    print("High Processed-Food Signal:", f"{processed_count / total * 100:.1f}%")
    print("High Sugary-Drink Signal:", f"{sugary_count / total * 100:.1f}%")
    print("Average Diversity Score:", round(sum(diversity_scores) / total, 2))

    return {
        "total_participants": total,
        "average_risk_score": sum(scores) / total,
        "highest_risk_score": max(scores),
        "low_risk_records": low_count,
        "moderate_risk_records": moderate_count,
        "higher_risk_records": high_count,
        "high_processed_food_percentage": processed_count / total * 100,
        "high_sugary_drink_percentage": sugary_count / total * 100,
        "average_diversity_score": sum(diversity_scores) / total
    }


def management_summary(food_data):
    print("\n========== MANAGEMENT SUMMARY ==========")

    summary = generate_summary(food_data)
    if summary is None:
        return

    risk_counts = risk_factor_frequency(food_data)
    averages = location_analysis(food_data)
    invalid_records = validate_dataset(food_data)

    highest_location = max(averages, key=averages.get)
    most_common_signal = max(risk_counts, key=risk_counts.get)

    print("\nTotal Participants:", summary["total_participants"])
    print("Average Risk Score:", round(summary["average_risk_score"], 2))
    print("Highest Risk Score:", summary["highest_risk_score"])
    print("High-Risk Records:", summary["higher_risk_records"])
    print("Moderate-Risk Records:", summary["moderate_risk_records"])
    print("Low-Risk Records:", summary["low_risk_records"])
    print("Most Common Risk Signal:", most_common_signal)
    print("Highest-Risk Location:", highest_location)
    print("Average Diversity Score:", round(summary["average_diversity_score"], 2))
    print("Invalid Records:", invalid_records)



# round() is used to reduce a decimal number to a specified number of decimal places.
# In this project, it can be used when displaying the average risk score, so that the result is easier to read.

# sum() adds numbers together.
# In this project, it can be used to calculate the total of several values, such as the total risk scores of participants.

# max() returns the highest value from a group of numbers.
# In this project, it can be used to identify the highest risk score among the participants.
```

```python
def analysis_menu(food_data):

    while True:

        print("\n========== ANALYSIS MENU ==========")
        print("1. Risk Analysis")
        print("2. Risk Factor Frequency")
        print("3. Location Analysis")
        print("4. Risk Ranking")
        print("5. Back to Main Menu")

        choice = input("Enter your choice: ")

        if choice == "1":

            show_risk_analysis(food_data)

        elif choice == "2":

            risk_factor_frequency(food_data)

        elif choice == "3":

            location_analysis(food_data)

        elif choice == "4":

            risk_ranking(food_data)

        elif choice == "5":

            break

        else:

            print("Invalid option. Please try again.")
```

```python
def main_menu(food_data):

    while True:

        print("\n========================================")
        print("MAIN MENU")
        print("========================================")
        print("1. View All Participants")
        print("2. Search Participant")
        print("3. Search by Location")
        print("4. Data Analysis")
        print("5. High-Risk Signals")
        print("6. Validate Dataset")
        print("7. Add Participant")
        print("8. Management Summary")
        print("9. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":

            view_records(food_data)

        elif choice == "2":

            search_participant(food_data)

        elif choice == "3":

            search_location(food_data)

        elif choice == "4":

            analysis_menu(food_data)

        elif choice == "5":

            show_high_risk_records(food_data)

        elif choice == "6":

            validate_dataset(food_data)

        elif choice == "7":

            add_record(food_data)

        elif choice == "8":

            generate_summary(food_data)
            highest_risk_participant(food_data)

        elif choice == "9":

            print("\nThank you for using the system.")
            break

        else:

            print("Invalid option. Please choose 1-9.")
```
